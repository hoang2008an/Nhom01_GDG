```python
from sklearnex import patch_sklearn

patch_sklearn()

import pandas as pd
from ipynb.fs.defs.preprocessing import get_data
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import RandomOverSampler, ADASYN, SMOTE
from util import summary
```

```python
X_train, X_test, y_train, y_test = get_data(use_val=False, test_size=0.3)
```

```python
X_train
```

```python
clf = Pipeline([("scaler", StandardScaler()), ("logreg", LogisticRegression())])
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
summary(y_test, y_pred)
```

```python
ros = RandomOverSampler(random_state=42, sampling_strategy=0.01)
X_resampled, y_resampled = ros.fit_resample(X_train, y_train)
clf = Pipeline([("scaler", StandardScaler()), ("logreg", LogisticRegression())])
clf.fit(X_resampled, y_resampled)
y_pred = clf.predict(X_test)
summary(y_test, y_pred)
```

```python
ros = SMOTE(random_state=42, sampling_strategy=0.01, k_neighbors=5)
X_resampled, y_resampled = ros.fit_resample(X_train, y_train)
clf = Pipeline([("scaler", StandardScaler()), ("logreg", LogisticRegression())])
clf.fit(X_resampled, y_resampled)
y_pred = clf.predict(X_test)
summary(y_test, y_pred)
```

```python
ros = ADASYN(random_state=42, sampling_strategy=0.01)
X_resampled, y_resampled = ros.fit_resample(X_train, y_train)
clf = Pipeline([("scaler", StandardScaler()), ("logreg", LogisticRegression())])
clf.fit(X_resampled, y_resampled)
y_pred = clf.predict(X_test)
summary(y_test, y_pred)
```
