```python
from sklearnex import patch_sklearn

patch_sklearn()

import pandas as pd
from ipynb.fs.defs.preprocessing import get_data
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from util import summary
```

```python
X_train, X_test, y_train, y_test = get_data(use_val=False, test_size=0.3)
```

```python
X_train
```

```python
clf = Pipeline([("scaler", StandardScaler()), ("logreg", LogisticRegression())])
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
summary(y_test, y_pred)
```

```python
probs = clf.predict_proba(X_test)[:, 1]
y_pred = (probs >= 0.2).astype(int)
summary(y_test, y_pred)
```
