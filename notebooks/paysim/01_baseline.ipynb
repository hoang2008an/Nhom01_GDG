# PaySim: Baseline Logistic Regression & Class Weight
Thực nghiệm mô hình cơ sở Logistic Regression trên PaySim với Feature Engineering chống rò rỉ dữ liệu (No-balance).


```
import pandas as pd
from ipynb.fs.defs.preprocessing import get_data
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from util import summary
```

```
X_train, X_test, y_train, y_test = get_data(sample_size=150000, test_size=0.3)
print("Train shape:", X_train.shape, "Fraud count:", int(y_train.sum()))
print("Test shape: ", X_test.shape, "Fraud count: ", int(y_test.sum()))
```

### 1. Logistic Regression Mặc Định (Không xử lý mất cân bằng)

```
pipe_base = Pipeline([("scaler", StandardScaler()), ("lr", LogisticRegression(max_iter=1000, random_state=42))])
pipe_base.fit(X_train, y_train)
y_pred_base = pipe_base.predict(X_test)
summary(y_test, y_pred_base)
```

### 2. Logistic Regression với Class Weight Balanced

```
pipe_bal = Pipeline([("scaler", StandardScaler()), ("lr", LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42))])
pipe_bal.fit(X_train, y_train)
y_pred_bal = pipe_bal.predict(X_test)
summary(y_test, y_pred_bal)
```
