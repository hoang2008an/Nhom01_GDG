# PaySim: Tree-Based Models (Random Forest, XGBoost, LightGBM)
So sánh các mô hình cây quyết định và Gradient Boosting với trọng số cân bằng `scale_pos_weight`.


```
import pandas as pd
from ipynb.fs.defs.preprocessing import get_data
from sklearn.ensemble import RandomForestClassifier
import xgboost as xgb
import lightgbm as lgb
from util import summary
```

```
X_train, X_test, y_train, y_test = get_data(sample_size=150000, test_size=0.3)
pos_weight = (len(y_train) - y_train.sum()) / y_train.sum()
print("scale_pos_weight:", round(pos_weight, 2))
```

### 1. Random Forest (Class Weight Balanced)

```
rf = RandomForestClassifier(n_estimators=100, max_depth=12, class_weight="balanced", random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
summary(y_test, y_pred_rf)
```

### 2. XGBoost (scale_pos_weight)

```
xgb_clf = xgb.XGBClassifier(
    n_estimators=150, max_depth=6, learning_rate=0.05,
    scale_pos_weight=pos_weight, eval_metric="aucpr",
    random_state=42, n_jobs=-1
)
xgb_clf.fit(X_train, y_train)
y_pred_xgb = xgb_clf.predict(X_test)
summary(y_test, y_pred_xgb)
```

### 3. LightGBM (scale_pos_weight)

```
lgb_clf = lgb.LGBMClassifier(
    n_estimators=150, num_leaves=63, learning_rate=0.05,
    scale_pos_weight=pos_weight, random_state=42, n_jobs=-1, verbose=-1
)
lgb_clf.fit(X_train, y_train)
y_pred_lgb = lgb_clf.predict(X_test)
summary(y_test, y_pred_lgb)
```
